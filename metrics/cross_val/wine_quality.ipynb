```python
import pandas as pd 
data = pd.read_csv("../datasets/wineQualityReds.csv")
data.tail()
```

Map the quality values from 0-10 to 0-5 and split the shuffled data into testing and validation

```python
quality_mapping = {
    3:0,
    4:1,
    5:2,
    6:3,
    7:4,
    8:5
}
print(data["quality"])
data["quality"] = data.quality.astype(int).map(quality_mapping)
print(data["quality"])
test_size = 1000

if len(data) > test_size:
    shuff_data = data.sample(frac = 1).reset_index(drop = True).copy()
    train = data.head(test_size)
    validation = data.tail(len(data)-test_size)
    print(f"Train: {train.shape}")
    print(f"Validation: {validation.shape}")
else:
    print("Test split cannot be larger than data set size.")


```

Use decision trees from sklearn and fit the model with depth 3 for train and validation

```python
from sklearn import tree, metrics

clf = tree.DecisionTreeClassifier(max_depth = 3)

cols = ['fixed.acidity', 'volatile.acidity', 'citric.acid',
       'residual.sugar', 'chlorides', 'free.sulfur.dioxide',
       'total.sulfur.dioxide', 'density', 'pH', 'sulphates', 'alcohol']
clf.fit(train[cols], train.quality)
train_predicitons = clf.predict(train[cols])
val_predictions = clf.predict(validation[cols])

train_acc = metrics.accuracy_score(
    train.quality, train_predicitons
)
val_acc = metrics.accuracy_score(
    validation.quality, val_predictions
)

print(f"Train Acc: {train_acc:.3f}, Val Acc: {val_acc:.3f}")
print(train["quality"].value_counts())
```
